# Fibre modes widget

Opens the PySide6 window over the catalog. Install the extra first if the import fails: `pip install -e ".[gui]"`.

Restart the kernel before running this cell again. The kernel has to own the Qt event loop; a window left open by an older run stays frozen.

Pick **analytical** for LG or HG, or **numerical** for a fibre and the radial or scalar solver. **Auto samples** sizes the solver grid from the wavelength and NA. Uncheck it to type `n_samples`. **Run** solves. **Viewport** shows the complex mode and the incoherent sum. Change the resample sample count and half-width, then **Resample**. The viewport uses that grid when a resample exists.

```python
import sys
from pathlib import Path

_start = Path.cwd().resolve()
for _folder in (_start, *_start.parents):
    if (_folder / "fibremodes" / "ui" / "window.py").is_file():
        _algorithms = _folder
        break
    if _folder.name == "fibremodes" and (_folder / "ui" / "window.py").is_file():
        _algorithms = _folder.parent
        break
    if (_folder / "algorithms" / "fibremodes" / "ui" / "window.py").is_file():
        _algorithms = _folder / "algorithms"
        break
else:
    raise RuntimeError("Could not find the fibremodes package from the notebook working directory.")
if str(_algorithms) not in sys.path:
    sys.path.insert(0, str(_algorithms))

from fibremodes.ui import launch

window = launch()
```
